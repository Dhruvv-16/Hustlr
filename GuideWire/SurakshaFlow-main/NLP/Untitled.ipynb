```python
AI = """According to the father of Artificial Intelligence, John McCarthy, it is "The science and engineering 
of making intellige Artificial Intelligence is a way of making a computer, a computer-controlled robot, or a 
software think intelligently, in the sim AI is accomplished by studying how human brain thinks, and how humans 
learn, decide, and work while trying to solve a problem, an I While exploiting the power of the computer systems, 
the curiosity of human, lead him to wonder, "Can a machine think and behave 1 Thus, the development of AI started
with the intention of creating similar intelligence in machines that we find and regard high To Create Expert 
Systems The systems which exhibit intelligent behavior, learn, demonstrate, explain, and advice its users. To 
Implement Human Intelligence in Machines Creating systems that understand, think, learn, and behave like humans.
Artificial intelligence is a science and technology based on disciplines such as Computer Science, Biology,
Psychology, Linguistic
Out of the following areas, one or multiple areas can contribute to build an intelligent system."""
```

```python
type(AI)
```

```python
from nltk.tokenize import word_tokenize
```

```python
AI_tokens = word_tokenize(AI)
AI_tokens
```

```python
len(AI_tokens)
```

```python
from nltk.probability import FreqDist
fdist = FreqDist()
```

```python
for word in AI_tokens:
    fdist[word.lower()]+=1
fdist
```

```python
len(fdist)
```

```python
fdist_top10 = fdist.most_common(10)
fdist_top10 
```

```python
string = "The best and most beautiful things in the world cannot be seen or even touched, they must be felt with the heart"
quotes_tokens = nltk.word_tokenize(string)
quotes_tokens
```

```python
quotes_bigrams = list(nltk.bigrams (quotes_tokens))
quotes_bigrams
```

```python
quotes_trigrams = list(nltk.trigrams (quotes_tokens))
quotes_trigrams
```

```python
quotes_ngrams = list(nltk.ngrams (quotes_tokens , 5))
quotes_ngrams
```

```python
from nltk.stem import PorterStemmer
pst= PorterStemmer()
```

```python
pst.stem("having")
```

```python
words_to_stem =["give", "giving", "given", "gave"]
for words in words_to_stem: 
    print (words+ ":" +pst.stem(words))
```

```python
from nltk.stem import LancasterStemmer
lst=LancasterStemmer()
for words in words_to_stem:
    print(words+ ":" +lst.stem(words))
```

```python
from nltk.stem import SnowballStemmer
sbst= SnowballStemmer('english')
```

```python
for words in words_to_stem:
    print(words+":"+sbst.stem(words))
```
